## 6. Demonstração do fluxo integrado

```
from PIL import Image
from src.config import DIR_IMAGENS
from src.pipeline import AssistenteResiduos
a = AssistenteResiduos()
img = Image.open(next((DIR_IMAGENS/'glass').iterdir())).convert('RGB')
display(img.resize((256, 192)))
for pergunta in ['qual a cor da lixeira pra isso?', 'preciso lavar antes?', 'onde posso descartar?']:
    r = a.responder(img, pergunta)
    print(f'> {pergunta}\n{r.decisao.acao}: {r.decisao.resposta}\nregras: {r.decisao.rastro}\n')
```
